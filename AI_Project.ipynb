```python
import os

print("Train folder contains:")
print(os.listdir("/content/ImageClassificationDataset/train"))

print("\nTest folder contains:")
print(os.listdir("/content/ImageClassificationDataset/test"))

```

```python
train_dir = "/content/ImageClassificationDataset/train" #just variables that store the location (path) of your folders.
test_dir  = "/content/ImageClassificationDataset/test"
```

```python
import tensorflow as tf
#helps Load images Build neural networks Train models Evaluate accuracy Make predictions Without importing it

img_size = (180, 180) #Sets the size to which all images will be resized (180×180).
batch_size = 32 #It will load just 32 images, train on them, then move to the next 32, and so on.

train_ds = tf.keras.preprocessing.image_dataset_from_directory(  #Reads images from folders of my computer.
    train_dir, #Path to training folder.
    image_size=img_size, #Resizes every image to (180×180
    batch_size=batch_size, #Loads images in batches of 32
    shuffle=True #Randomly mixes images so the model learns better
)

test_ds = tf.keras.preprocessing.image_dataset_from_directory( #reads all images in  folder, label it0,1,2, , resize, gruop in to batches,
    test_dir, #contain test images
    image_size=img_size,
    batch_size=batch_size,
    shuffle=False #Do NOT mix the order of test images Because test data should always be:evaluated in a fixed ordernot randomized not altered
)
```

```python
from tensorflow.keras import layers, models #layer: building blocks of neural networks (Conv2D, Dense, MaxPooling, etc.)
#model: combine layers to create a complete neural network

num_classes = 5 # 5 type of flower to neuron

model = models.Sequential([ #Sequential model = layers stacked one after another.
    layers.Rescaling(1./255, input_shape=img_size + (3,)), #Divides all pixel values by 255 to normalize them between 0 and 1., input shape: expects images of size (180, 180) with 3 color channels (RGB).

    layers.Conv2D(32, (3,3), activation='relu'), #Conv2 Ddetects features (edges, shapes, colors learn them) in images. #32 → number of filters (features it will learn) (3,3) → size of the filter (kernel)
    layers.MaxPooling2D(), #reduce computation and prevent overfitting.

    layers.Conv2D(64, (3,3), activation='relu'),#Number of filters increases to 64 → can learn more patterns.petal shapes, curves).
    layers.MaxPooling2D(),

    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(),

    layers.Flatten(), #Converts 2D feature maps from previous layers into 1D array
    layers.Dense(128, activation='relu'), #Learns complex combinations of features detected earlier.128 neurons → allows the model to combine features in many ways.
    layers.Dense(num_classes, activation='softmax') #softmax activation → converts raw numbers into probabilities that sum to 1.
])

model.compile(
    optimizer='adam', #decides how the model updates weights to learn.
    loss='sparse_categorical_crossentropy',#measures how wrong predictions are
    metrics=['accuracy']#track accuracy during training.
)

model.summary()
#Param : learn number in layer if there is weight, max pooling does not gave weight. Formula =(hight and weightand input channel +1)filter
#the amount of memory needed to store the model’s learnable weights
#con2d=input-3/1+1
#max ()-2/2+1
```

```python
history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=10

)
#Epoch is one complete pass through entire train data file, if its 1/10 goes from 1 t 10
#Loss=−N1​i=1∑N​c=1∑C​yi,c​log(y^​i,c​)
#Accuracy=Total number of samples/Number of correct predictions​
```

```python
loss, acc = model.evaluate(test_ds)
print(f"Test Accuracy: {acc:.3f}")
print(f"Test Loss: {loss:.3f}")
#how many prediction was correct and wrong
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ---- 1.TRUE LABELS ----
y_true = np.concatenate([y for x, y in test_ds], axis=0)

# ---- 2. PREDICTED LABELS ----
y_pred = np.argmax(model.predict(test_ds), axis=1)

# Get class names from the dataset
class_names = test_ds.class_names

# ---- 3. CLASSIFICATION REPORT ----
print("Classification Report:\n")
print(classification_report(y_true, y_pred, target_names=class_names))

# ---- 4. CONFUSION MATRIX ----
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", #annot=True → writes numbers in the squares. xticklabels / yticklabels → uses your class names.
            xticklabels=class_names,
            yticklabels=class_names)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()



#Precision: How many predicted as class X were actually X.

#Recall: How many actual class X were correctly predicted.

#F1-score: Harmonic mean of precision and recall.

#Support: Number of true samples in that class.
#Macro F1 = 0.34, meaning all classes perform badly & Weighted F1 = 0.35, meaning even the larger classes are not predicted well
#macro avg recall=Rdaisy​+Rdandelion​+Rrose​+Rsunflower​+Rtulip/5​​
#weighted recall=∑(class recall×number of samples)/total samples
```
